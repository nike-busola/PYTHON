cufflinks is a python library that connects plotly with pandas so that we can create charts directly on dataframes. 
import init_notebook_mode-- enables the offline mode for plotting locally in the jupyter noteboook

```python
!pip install plotly
!pip install cufflinks
!pip install chart_studio
from plotly.offline import init_notebook_mode
import cufflinks as cf
cf.go_offline()
import pandas as pd
```

```python
pro = pd.read_excel("products summary.xlsx","sales details")
pro
```

```python
a = pro.groupby(["ZONE"]).sum(numeric_only=True)["TOTAL"].reset_index()
a
```

```python
#Original Cufflinks approach
a.iplot(kind="bar",x="ZONE",y="TOTAL",text="TOTAL",
       title="TOTAL SALES BY ZONES",
       yaxis_title="TOTAL",xaxis_title="ZONE")
```

```python
b = pro.groupby(["PRODUCT"]).sum(numeric_only=True)["UNIT PRICE"].reset_index()
b
```

```python
#Original Cufflinks approach
b.iplot(kind="pie",labels="PRODUCT",values="UNIT PRICE",text="TOTAL",
       title="TOTAL UNIT PRICE BY PRODUCT")
```

```python
c = pro.groupby(["MONTH"]).sum(numeric_only=True)["TOTAL"].reset_index()
c
```

```python
#Original Cufflinks approach
c.iplot(kind="pie",labels="MONTH",values="TOTAL",hole=0.5,
       title="TOTAL SALES BY MONTH")
```

```python
a = pro.groupby(["ZONE"]).sum(numeric_only=True)["TOTAL"].reset_index()
a
```

```python
import plotly.express as px

fig = px.bar(
    a,
    x="ZONE",
    y="TOTAL",
    text="TOTAL",
    title="TOTAL SALES BY ZONES"
)

fig.update_layout(
    xaxis_title="ZONE",
    yaxis_title="TOTAL"
)

fig.show()
```

```python
b = pro.groupby(["PRODUCT"]).sum(numeric_only=True)["UNIT PRICE"].reset_index()
b
```

```python
fig = px.pie(
    b,
    names="PRODUCT",
    values="UNIT PRICE",
    title="TOTAL UNIT PRICE BY PRODUCT"
)

fig.update_traces(textinfo="label+value")

fig.show()
```

```python
c = pro.groupby(["MONTH"]).sum(numeric_only=True)["TOTAL"].reset_index()
c
```

```python
fig = px.pie(
    c,
    names="MONTH",
    values="TOTAL",
    hole=0.5,
    title="TOTAL SALES BY MONTH"
)

fig.show()
```
