Seaborn is also a visualization library and it uses the matplotlib background

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pivot = pd.read_excel("pivot table.xlsx","Practice",header=[2])
pivot
```

```python
del pivot["Unnamed: 0"]
```

```python
pivot
```

```python
plt.figure(figsize=[25,10])
sns.barplot(x="Item_Type",y="Item_Outlet_Sales",data=pivot)
plt.title("TOTAL SALES BY ITEM_TYPE")
plt.show()
```

```python
plt.figure(figsize=[10,8])
sns.boxplot(x="Outlet_Type",y="Item_Outlet_Sales",data=pivot)
plt.title("TOTAL SALES BY OUTLET_TYPE")
plt.show()
```

```python
plt.figure(figsize=[10,8])
sns.barplot(x="Outlet_Location_Type",y="Item_Outlet_Sales",data=pivot)
plt.title("TOTAL SALES BY OUTLET_LOCATION_TYPE")
plt.show()
```

```python
plt.figure(figsize=[25,10])
sns.boxplot(x="Item_Fat_Content",y="Item_Outlet_Sales",data=pivot)
plt.title("TOTAL SALES BY ITEM_FAT_CONTENT")
plt.show()
```

```python
""""with subplot() function you can draw multiple plots in one figure
The subplot() function takes three arguments that describes the layout of
the figure.
The layout is organized in rows and columns, which are represented by the first
and second argument.
The third argument represents the index of the current plot.
plt.subplot(1, 2, 1)
#the figure has 1 row, 2 columns, and this plot is the first plot.
plt.subplot(1, 2, 2)
#the figure has 1 row, 2 columns, and this plot is the second plot."""
```

```python
plt.figure(figsize=[35,10])

plt.subplot(2,2,1)
sns.barplot(x="Item_Type",y="Item_Outlet_Sales",data=pivot)

plt.subplot(2,2,2)
sns.boxplot(x="Outlet_Type",y="Item_Outlet_Sales",data=pivot)

plt.subplot(2,2,3)
sns.barplot(x="Outlet_Location_Type",y="Item_Outlet_Sales",data=pivot)

plt.subplot(2,2,4)
sns.boxplot(x="Item_Fat_Content",y="Item_Outlet_Sales",data=pivot)

plt.show()
```
